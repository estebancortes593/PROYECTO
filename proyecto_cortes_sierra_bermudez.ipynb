# Planificador de Itinerarios Disney World

Ejecuta las celdas **en orden** (o *Entorno de ejecución → Ejecutar todo*):

1. **Preparación**: carga los datos.
2. **Parque y año**: muestra el catálogo de atracciones con su espera.
3. **Itinerario**: escribe los números de tu ruta, ajusta tus horas y mira el reporte.

Cada vez que cambies una casilla, vuelve a presionar ▶ en esa celda.

```python
#@title 1 · Preparación de datos (ejecuta esta celda primero) { display-mode: "form" }
#@markdown ### ⚙️ Preparación
#@markdown Esta celda carga los datos desde Drive y deja listas las funciones del planificador.


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML
from google.colab import drive

drive.mount('/content/drive')
ruta_archivo = '/content/drive/MyDrive/Programación/ride_stats.csv'
df = pd.read_csv(ruta_archivo)

df_limpio = df[df['is_archived'] == 0].copy()
df_limpio['year'] = pd.to_numeric(df_limpio['year'], errors='coerce')


def obtener_atracciones_parque(dataframe, parque, anio):
    filtro = (dataframe['park_name'] == parque) & (dataframe['year'] == anio)
    sub_df = dataframe[filtro]

    if sub_df.empty:
        return pd.DataFrame()

    tabla = sub_df.pivot_table(
        index='ride_name',
        columns='metric',
        values='minutes',
        aggfunc='mean'
    ).reset_index()

    return tabla.sort_values(by='ride_avg_wait', ascending=False).reset_index(drop=True)


def simular_itinerario(lista_seleccionadas, minutos_disponibles, tiempo_recorrido=15):

    atendidas_normal = []
    pendientes_normal = []
    atendidas_max = []
    pendientes_max = []

    tiempo_acum_normal = 0
    tiempo_acum_max = 0

    for atraccion in lista_seleccionadas:
        nombre = atraccion['nombre']
        # Sumamos los minutos de fila + minutos de duración/caminata entre atracciones
        costo_normal = atraccion['espera_promedio'] + tiempo_recorrido
        costo_max = atraccion['espera_maxima'] + tiempo_recorrido

        # Evaluación en Escenario Normal
        if tiempo_acum_normal + costo_normal <= minutos_disponibles:
            tiempo_acum_normal += costo_normal
            atendidas_normal.append((nombre, atraccion['espera_promedio']))
        else:
            pendientes_normal.append((nombre, atraccion['espera_promedio']))

        # Evaluación en Escenario de Alta Demanda (Máximo)
        if tiempo_acum_max + costo_max <= minutos_disponibles:
            tiempo_acum_max += costo_max
            atendidas_max.append((nombre, atraccion['espera_maxima']))
        else:
            pendientes_max.append((nombre, atraccion['espera_maxima']))

    return {
        'atendidas_normal': atendidas_normal,
        'pendientes_normal': pendientes_normal,
        'tiempo_normal': tiempo_acum_normal,
        'atendidas_max': atendidas_max,
        'pendientes_max': pendientes_max,
        'tiempo_max': tiempo_acum_max
    }


def tarjeta(titulo, valor, detalle, fondo, borde, ancho="25%"):
    # Tarjeta de resultado en HTML (background, border-left, padding, border-radius)
    return f"""
    <div style="display:inline-block; vertical-align:top; width:{ancho}; margin:6px;
    background:{fondo}; border-left:6px solid {borde};
    padding:14px 20px; border-radius:6px; font-family:sans-serif; color:#1f2a37">
    <div style="font-size:14px">{titulo}</div>
    <b style="font-size:34px">{valor}</b>
    <div style="font-size:13px">{detalle}</div>
    </div>"""


display(HTML(tarjeta(
    "✅ Datos listos",
    f"{len(df_limpio):,.0f} registros",
    f"Parques: {', '.join(df_limpio['park_name'].unique())}",
    "#eaf6ee", "#2e8b57", ancho="85%"
)))
```

```python
#@title 2 · Elige tu parque y año { display-mode: "form" }
#@markdown ### 🏰 Planificador de Itinerarios Disney
#@markdown Elige el **año** de referencia y el **parque** que visitarás. Luego presiona ▶.

anio = 2021 #@param {type:"slider", min:2015, max:2026, step:1}
parque = "Epcot" #@param ["Epcot", "Magic Kingdom", "Hollywood Studios", "Animal Kingdom"]

catalogo = obtener_atracciones_parque(df_limpio, parque, anio)

if catalogo.empty:
    display(HTML(tarjeta(
        "⚠️ Sin datos",
        "No hay registros",
        f"No se encontraron datos para {parque} en {anio}. Prueba con otro año o parque.",
        "#fdecea", "#c0392b", ancho="85%"
    )))
else:
    espera_parque = catalogo['ride_avg_wait'].mean()
    mas_larga = catalogo.iloc[0]

    display(HTML(
        tarjeta("🏰 Parque", parque, f"Año de referencia: {anio}", "#eef4fc", "#2a78d6")
        + tarjeta("🎢 Atracciones activas", f"{len(catalogo)}", "Disponibles para tu ruta", "#f3eefc", "#7b4fc9")
        + tarjeta("⏱️ Espera promedio", f"{espera_parque:.0f} min",
                  f"La más larga: {mas_larga['ride_name']} ({mas_larga['ride_avg_wait']:.0f} min)",
                  "#fdf0e6", "#d95f02")
    ))

    catalogo_vista = catalogo[['ride_name', 'ride_avg_wait', 'ride_max_wait']].copy()
    catalogo_vista.columns = ['Atracción', 'Espera promedio (min)', 'Espera máxima (min)']
    catalogo_vista.index = range(1, len(catalogo_vista) + 1)

    display(catalogo_vista.style
        .format({'Espera promedio (min)': '{:.0f}', 'Espera máxima (min)': '{:.0f}'})
        .background_gradient(subset=['Espera promedio (min)', 'Espera máxima (min)'], cmap='Blues')
        .set_caption(f'Atracciones de {parque} ({anio}) · Usa el número de la izquierda en la celda 3'))
```

```python
#@title 3 · Arma tu itinerario y mira el reporte { display-mode: "form" }
#@markdown ### 🗺️ Tu ruta del día
#@markdown Escribe los **números** de las atracciones de la tabla de la celda 2,
#@markdown separados por comas y en el **orden** en que las visitarás.
atracciones = "1, 2, 3, 6, 7, 11, 10, 4" #@param {type:"string"}

#@markdown ¿Cuántas **horas** tienes en el parque para subir a atracciones?
horas = 7 #@param {type:"slider", min:1, max:14, step:0.5}

#@markdown Minutos de **caminata y duración** entre una atracción y la siguiente:
tiempo_recorrido = 5 #@param {type:"slider", min:5, max:30, step:5}

minutos_disp = horas * 60

# Convertimos el texto "1, 2, 3" en la lista de atracciones elegidas
itinerario_usuario = []
invalidos = []
for parte in atracciones.split(","):
    parte = parte.strip()
    if parte.isdigit() and 1 <= int(parte) <= len(catalogo):
        fila_sel = catalogo.iloc[int(parte) - 1]
        itinerario_usuario.append({
            'nombre': fila_sel['ride_name'],
            'espera_promedio': fila_sel['ride_avg_wait'],
            'espera_maxima': fila_sel['ride_max_wait']
        })
    elif parte != "":
        invalidos.append(parte)

if catalogo.empty:
    display(HTML(tarjeta(
        "⚠️ Primero elige un parque con datos",
        "Sin catálogo",
        "Vuelve a la celda 2 y escoge un año y parque que tengan registros.",
        "#fdecea", "#c0392b", ancho="85%"
    )))
elif len(itinerario_usuario) == 0:
    display(HTML(tarjeta(
        "⚠️ Tu ruta está vacía",
        "0 atracciones",
        f"Escribe al menos un número entre 1 y {len(catalogo)}, por ejemplo: 1, 4, 7",
        "#fdecea", "#c0392b", ancho="85%"
    )))
else:
    if len(invalidos) > 0:
        display(HTML(tarjeta(
            "⚠️ Ignoré algunos valores",
            f"{len(invalidos)} inválido(s)",
            f"No están en la lista (1 a {len(catalogo)}): {', '.join(invalidos)}",
            "#fff8e1", "#e0a800", ancho="85%"
        )))

    resultado = simular_itinerario(itinerario_usuario, minutos_disp, tiempo_recorrido)
    total = len(itinerario_usuario)

    fila_total_max = sum([x['espera_maxima'] for x in itinerario_usuario])
    ahorro_minutos = fila_total_max * 0.75
    conviene_pase = len(resultado['pendientes_max']) > 0

    # ---------- 1. Tarjetas con las cifras clave ----------
    display(HTML(f"""
    <div style="font-family:sans-serif; font-size:22px; margin:10px 6px 4px 6px">
    <b>Reporte de viabilidad · {parque} ({anio})</b> &nbsp;·&nbsp; {minutos_disp:,.0f} minutos disponibles
    </div>"""))

    display(HTML(
        tarjeta("🙂 Escenario normal", f"{len(resultado['atendidas_normal'])} de {total}",
                f"Tiempo requerido: {resultado['tiempo_normal']:,.0f} min "
                f"({resultado['tiempo_normal'] / minutos_disp:.0%} de tu tiempo)",
                "#eef4fc", "#2b5c8f")
        + tarjeta("🔥 Alta demanda", f"{len(resultado['atendidas_max'])} de {total}",
                  f"Tiempo requerido: {resultado['tiempo_max']:,.0f} min "
                  f"({resultado['tiempo_max'] / minutos_disp:.0%} de tu tiempo)",
                  "#fdf0e6", "#d95f02")
        + tarjeta("⚡ Ahorro con Pase Rápido", f"~{ahorro_minutos:,.0f} min",
                  f"Unas {ahorro_minutos / 60:.1f} horas menos de fila en un día lleno",
                  "#f3eefc", "#7b4fc9")
    ))

    if conviene_pase:
        pendientes = ', '.join([x[0] for x in resultado['pendientes_max']])
        display(HTML(tarjeta(
            "Veredicto · Pase Rápido / Lightning Lane",
            "¡SÍ te conviene comprarlo!",
            f"En alta demanda no alcanzas: {pendientes}.",
            "#fdecea", "#c0392b", ancho="85%"
        )))
    else:
        display(HTML(tarjeta(
            "Veredicto · Pase Rápido / Lightning Lane",
            "No necesitas pagarlo",
            "Tu tiempo alcanza incluso en el escenario de alta demanda.",
            "#eaf6ee", "#2e8b57", ancho="85%"
        )))

    # ---------- 2. Tres gráficos en un solo lienzo ----------
    nombres = [item['nombre'][:22] + '...' if len(item['nombre']) > 22 else item['nombre'] for item in itinerario_usuario]
    promedios = [item['espera_promedio'] for item in itinerario_usuario]
    maximos = [item['espera_maxima'] for item in itinerario_usuario]

    paradas = [str(i) for i in range(1, total + 1)]
    acum_normal = np.cumsum([p + tiempo_recorrido for p in promedios])
    acum_max = np.cumsum([m + tiempo_recorrido for m in maximos])

    fila_hecha = sum([x[1] for x in resultado['atendidas_max']])
    recorrido_hecho = len(resultado['atendidas_max']) * tiempo_recorrido
    libre = minutos_disp - resultado['tiempo_max']
    partes_dia = [(f"Fila\n{fila_hecha:,.0f} min", fila_hecha, '#d95f02'),
                  (f"Caminata\n{recorrido_hecho:,.0f} min", recorrido_hecho, '#7b4fc9'),
                  (f"Tiempo libre\n{libre:,.0f} min", libre, '#2e8b57')]
    partes_dia = [p for p in partes_dia if p[1] > 0]

    x = np.arange(total)
    ancho = 0.35

    fig, axs = plt.subplots(1, 3, figsize=(17, 5))

    axs[0].bar(x - ancho/2, promedios, ancho, label='Normal (promedio)', color='#2b5c8f')
    axs[0].bar(x + ancho/2, maximos, ancho, label='Alta demanda (máximo)', color='#d95f02')
    axs[0].set_title('Minutos de fila por atracción')
    axs[0].set_ylabel('Minutos de fila')
    axs[0].set_xticks(x)
    axs[0].set_xticklabels(nombres, rotation=30, ha='right')
    axs[0].legend()

    axs[1].plot(paradas, acum_normal, marker='o', label='Normal', color='#2b5c8f')
    axs[1].plot(paradas, acum_max, marker='o', label='Alta demanda', color='#d95f02')
    axs[1].plot(paradas, [minutos_disp] * total, linestyle='--', label='Tu tiempo disponible', color='#555555')
    axs[1].set_title('Tiempo acumulado si haces toda la ruta')
    axs[1].set_xlabel('Parada de la ruta')
    axs[1].set_ylabel('Minutos')
    axs[1].legend()

    axs[2].pie([p[1] for p in partes_dia], labels=[p[0] for p in partes_dia],
               colors=[p[2] for p in partes_dia])
    axs[2].set_title('Así se va tu día (alta demanda)')

    plt.tight_layout()
    plt.show()

    # ---------- 3. Tabla con estilo ----------
    quedan_normal = [x[0] for x in resultado['atendidas_normal']]
    quedan_max = [x[0] for x in resultado['atendidas_max']]
    estado_normal = []
    estado_max = []
    for item in itinerario_usuario:
        if item['nombre'] in quedan_normal:
            estado_normal.append('✅ Sí')
            quedan_normal.remove(item['nombre'])
        else:
            estado_normal.append('⏳ Pendiente')
        if item['nombre'] in quedan_max:
            estado_max.append('✅ Sí')
            quedan_max.remove(item['nombre'])
        else:
            estado_max.append('⏳ Pendiente')

    detalle = pd.DataFrame({
        'Atracción': [item['nombre'] for item in itinerario_usuario],
        'Fila promedio (min)': promedios,
        'Fila máxima (min)': maximos,
        '¿Alcanzas? (normal)': estado_normal,
        '¿Alcanzas? (alta demanda)': estado_max
    })
    detalle.index = range(1, total + 1)

    display(detalle.style
        .format({'Fila promedio (min)': '{:.0f}', 'Fila máxima (min)': '{:.0f}'})
        .background_gradient(subset=['Fila máxima (min)'], cmap='Oranges')
        .set_caption('Detalle de tu ruta · Más oscuro = más fila'))
```
